```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

sns.set_theme(style="darkgrid")
```

```python
#load the simple dataset
df = pd.read_csv("hours and marks.csv")
print("Dataset:")
print(df.head())

#visualize the data
plt.figure(figsize = (8,5))
plt.scatter(df['hours'], df['pass'], c=df['pass'], cmap='plasma',edgecolor='k')
plt.xlabel('Hours Studied')
plt.ylabel('Pass (1) / Fail (0)')
plt.title('Hours Studied vs Pass/Fail')
plt.show()
```

```python
#prepare features and target
x = df[['hours']] #input 
y = df['pass']   #output

model = LogisticRegression()
model.fit(x,y)

print("model trained successfully")
```

```python
#visualizing the sigmoid curve(logistic regression curve)
plt.figure(figsize=(10, 6))
plt.scatter(df['hours'], df['pass'], c=df['pass'], cmap='bwr', label='Data Points')

#generate a range of values for the curve
x_test_range = np.linspace(0,10,100).reshape(-1,1)
y_prob_range = model.predict_proba(x_test_range)[:,1]

plt.plot(x_test_range, y_prob_range, color='green',linewidth=3, label='Logistic Regression Curve')
plt.axhline(y=0.5, color='gray', linestyle='--', label='Decision Boundary (0.5)')
plt.xlabel('Hours Studied')
plt.ylabel('Probability of Passing')
plt.title('Logistic Regression Curve')
plt.legend()
plt.show()
```

```python
try:
    #get user input for prediction
    user_hours=float(input("Enter study hours(0-10): "))
    
    #x_user input for model
    x_user=np.array([[user_hours]])

    #make prediction

    prediction=model.predict(x_user)[0]
    probability=model.predict_proba(x_user)[0][1]

    #output the result
    if prediction==1:
        print(f"\n Result pass")
        print(f"Confidence: {probability*100:.2f}")
    else:
        print(f"\n Result fail")
        print(f"Confidence: {(1-probability)*100:.2f}")

except ValueError:
    print("Invalid input. Please enter a number between 0 and 10.")
```
